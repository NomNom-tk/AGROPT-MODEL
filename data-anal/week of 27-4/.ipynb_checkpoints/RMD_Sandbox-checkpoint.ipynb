```r
# parquet file generation (RUN ONCE) 27/7/26
source("./framework_analysis.R")
# 2. Extract active CSV paths from the master container object
target_paths <- c(
  run_configs$lhs_main$batch$v1$path,
  run_configs$lhs_main$agent$v1$path,
  run_configs$lhs_main$interaction$v1$path,
  run_configs$ga_main$batch$v1$path,
  run_configs$ga_main$agent$v1$path,
  run_configs$ga_main$interaction$v1$path,
  "./data/data_complete_anonymised.csv"
)

# Filter out NULLs or missing file entries
target_paths <- unique(target_paths[!is.null(target_paths) & file.exists(target_paths)])



for (i in seq_along(target_paths)) {
  f_path <- target_paths[i]
  p_path <- sub("\\.csv$", ".parquet", f_path, ignore.case = TRUE)

  message(sprintf("Found %d target CSV files. Generating fresh Parquet caches...", length(target_paths)))

  # csv stream using arrow
  csv_stream <- read_csv_arrow(
    f_path,
    as_data_frame = FALSE)

  write_parquet(csv_stream, p_path)

  rm(csv_stream)
  gc()
}

message("we are finally done!")


# # 4. Read each CSV via read_clean() to build fresh parquet files
# for (i in seq_along(target_paths)) {
#   f_path <- target_paths[i]
#   cat(sprintf("[%d/%d] Processing & Caching: %s\n", i, length(target_paths), basename(f_path)))
  
#   tryCatch({
#     read_clean(f_path)
#   }, error = function(e) {
#     cat(sprintf("  -> ERROR on %s: %s\n", basename(f_path), e$message))
#   })
# }

message("\nAll Parquet cache files generated cleanly! You are ready to export.")
```

```r
# seed audit (do lhs and ga use the same seeds? if so, use a paired t test to determine relative improvement)

# cat message
cat("=== RUN CONFIG SEED AUDIT ===\n")

lhs_batch_df <- lhs_outputs$inputs$raw
ga_batch_df <- ga_outputs$inputs$raw

# helper function to identify columns with seed
find_seed_col <- function(df) {
  cols <- colnames(df)
  match <- cols[grepl("seed", cols, ignore.case = TRUE)]
  return(match)
}

lhs_seed_col <- find_seed_col(lhs_batch_df)
ga_seed_col <- find_seed_col(ga_batch_df)

if (length(lhs_seed_col) > 0) {
  lhs_seeds <- unique(lhs_batch_df[[lhs_seed_col]])
  cat("LHS seeds found:", paste(lhs_seeds, collapse = ", "), "\n")
} else {
  cat("Could not find LHS seed column")
  print(head(colnames(lhs_batch_df), 10))
}

if (length(ga_seed_col) > 0) {
  ga_seeds <- unique(ga_batch_df[[ga_seed_col]])
  cat("GA seeds found:", paste(ga_seeds, collapse = ", "), "\n")
} else {
  cat("Could not find GA seed column")
  print(head(colnames(ga_batch_df), 10))
}

# check for structural alignment
if (length(lhs_seed_col) > 0 && length(ga_seed_col) > 0) {
  cat("\n--- Conclusion ---\n")
  if (length(lhs_seeds) == length(ga_seeds) && all(lhs_seeds == ga_seeds)) {
    cat("Seeds match perfectly: proceed with paired t test. \n")
  } else {
    cat("Seeds do not match: proceed with independent t test. \n")
  }
}
```

```r
# GA-LHS ranking test (t test)
# extract mae values
lhs_vector <- lhs_outputs$inputs$raw$mae
ga_vector <- ga_outputs$inputs$raw$mae

# run welch's t test
defensive_ttest <- t.test(ga_vector, lhs_vector, paired = FALSE, alternative = "two.sided")

# print results
print(defensive_ttest)
```
