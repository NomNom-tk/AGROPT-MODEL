```r
# h3a check

# df empir test rebuild
df_empir_test <- ga_bundle$sim_val$df_ag %>%
    collect() %>%
    distinct(agent_id, debate_label, .keep_all = TRUE) %>%
    select(agent_id, debate_label, initial_opinion, final_attitude, current_condition)

mlm_bench_h3 <- lme4::lmer(final_attitude ~ initial_opinion + (1 | debate_label),
                           data = ga_bundle$sim_inputs$df_ag %>% collect() %>%
                           distinct(agent_id, debate_label, .keep_all = TRUE))

df_empir_test$mlm_pred <- predict(mlm_bench_h3, newdata = df_empir_test,
                                        allow.new.levels = TRUE)

df_empir_test$mlm_error <- abs(df_empir_test$final_attitude - df_empir_test$mlm_pred)
```

```r
# H3a set up 24/8/26
    df_h3a <- ga_bundle$sim_val$df_ag %>%
      collect() %>%
      distinct(agent_id, selected_debate_id, seed, model_type, .keep_all = TRUE) %>%
      mutate(abm_error = abs(final_attitude - opinion)) %>%
      left_join(df_empir_test %>% select(agent_id, debate_label, mlm_error), by = c("debate_label", "agent_id")) %>%
      mutate(model_type = relevel(factor(model_type), ref = "no_change"))

    log_step("Set up H3a df and starting lmer call...")
    mlm_h3a <- lme4::lmer(abm_error ~ model_type + ( 1 | debate_label),
                         data = df_h3a)
    summary(mlm_h3a)
```

```r
df_h3b <- ga_bundle$sim_val$df_batch %>%
      group_by(design_cell, debate_label, model_type) %>%
      summarize(abm_mae = mean(mae, na.rm = TRUE), .groups = "drop")

    # mlm debate level mae set up
    mlm_debate_mae <- df_empir_test %>%
      group_by(debate_label) %>%
      summarize(mlm_mae = mean(mlm_error, na.rm = TRUE), .groups = "drop")

    # No change mae extraction prior ot left_join
    nc_mae <- df_h3b %>%
      filter(model_type == "no_change") %>%
      select(debate_label, nc_mae = abm_mae)

    print(nc_mae)
    
    # join on debate_label to perform sign tests
    df_h3b <- df_h3b %>%
      filter(model_type != "no_change") %>%
      left_join(nc_mae, by = "debate_label") %>%
      left_join(mlm_debate_mae, by = "debate_label")

    # two sided sign tests set up
    ## ABM < NC across 12 debates
    abm_vs_nc <- df_h3b %>%
      group_by(design_cell) %>%
      summarize(
        wins = sum(abm_mae < nc_mae, na.rm = TRUE),
        n = n(),
        mean_diff = mean(abm_mae - nc_mae,na.rm = TRUE),
        p = binom.test(wins, n, 0.5)$p.value,
        .groups = "drop")

    ## ABM < MLM MAE across 12 debates
    abm_vs_mlm <- df_h3b %>%
      group_by(design_cell) %>%
      summarize(
        wins = sum(abm_mae < mlm_mae, na.rm = TRUE),
        n = n(),
        mean_diff = mean(abm_mae - mlm_mae, na.rm = TRUE),
        p = binom.test(wins, n, 0.5)$p.value,
        .groups = "drop")
```

```r
# seed audit (do lhs and ga use the same seeds? if so, use a paired t test to determine relative improvement)

# cat message
cat("=== RUN CONFIG SEED AUDIT ===\n")

lhs_batch_df <- lhs_outputs$inputs$raw
ga_batch_df <- ga_outputs$inputs$raw

# helper function to identify columns with seed
find_seed_col <- function(df) {
  cols <- colnames(df)
  match <- cols[grepl("seed", cols, ignore.case = TRUE)]
  return(match)
}

lhs_seed_col <- find_seed_col(lhs_batch_df)
ga_seed_col <- find_seed_col(ga_batch_df)

if (length(lhs_seed_col) > 0) {
  lhs_seeds <- unique(lhs_batch_df[[lhs_seed_col]])
  cat("LHS seeds found:", paste(lhs_seeds, collapse = ", "), "\n")
} else {
  cat("Could not find LHS seed column")
  print(head(colnames(lhs_batch_df), 10))
}

if (length(ga_seed_col) > 0) {
  ga_seeds <- unique(ga_batch_df[[ga_seed_col]])
  cat("GA seeds found:", paste(ga_seeds, collapse = ", "), "\n")
} else {
  cat("Could not find GA seed column")
  print(head(colnames(ga_batch_df), 10))
}

# check for structural alignment
if (length(lhs_seed_col) > 0 && length(ga_seed_col) > 0) {
  cat("\n--- Conclusion ---\n")
  if (length(lhs_seeds) == length(ga_seeds) && all(lhs_seeds == ga_seeds)) {
    cat("Seeds match perfectly: proceed with paired t test. \n")
  } else {
    cat("Seeds do not match: proceed with independent t test. \n")
  }
}
```
